```python
class Student():
    def __init__(self,name,subOne,subTwo,subThree):
        self.name = name
        self.subOne = subOne
        self.subTwo = subTwo
        self.subThree = subThree
        
    def  averageMarks(self):
        avg = (self.subOne+self.subTwo+self.subThree) /3
        print(f"Student name is {self.name} average score is {avg} ")

    @staticmethod     # static method agr hame simply print karwana hai to hum staticmethod use karte hain isme self likhne ki need nahi hoti
    def hello():
        print("Hello World")
        
    
        
    

obj = Student("Moiz",78,98,45)
obj.averageMarks()
obj.name = "Ahmed"
obj.averageMarks()
obj.hello()
```

```python
# Example (Abstraction)
# Hiding the implimintation detailed of a class and only showing the essential features to the user
class Car():
    def __init__(self):
        self.acc = False
        self.brk = False
        self.clutch = False

    def start(self):
        self.acc = True
        self.clutch = True
        print("Start...")

car1 = Car()
car1.start()
```
